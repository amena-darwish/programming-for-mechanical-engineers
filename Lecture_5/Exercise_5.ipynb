## Why use lists and dictionaries?

A **list** is useful when you want to store a sequence of values:

```python
horsepower = [110, 110, 93, 110, ...]
```

A **dictionary** is useful when one value identifies another value:

```python
cars["Honda Civic"]
```

A dictionary is also useful for grouping:

```python
cylinder_groups[6]
```



## Lists vs. Dictionaries

A **list** stores values in order. Each value has a **position**, called an index.

```python
cars = ["Volvo", "Toyota", "Honda"]

print(cars[1])
```

Output:

```text
Toyota
```

If we know a value but do not know its position, Python has to search through the list:

```python
position = cars.index("Honda")

print(position)
```

Output:

```text
2
```

So with a list, we normally access data using its **position**.


A **dictionary** works differently. Think of it like an old **telephone book**.

In a telephone book, names are organized by letters:

```text
A → Anna
J → John
S → Sara
```

If you know the person's **name**, you can go to the correct letter and find the phone number quickly.

For example:

```python
phone_book = {
    "Anna": "12345",
    "John": "67890",
    "Sara": "54321"
}
```

The person's name is the **key** and the phone number is the **value**:

```text
key        → value
"Anna"     → "12345"
"John"     → "67890"
"Sara"     → "54321"
```

If you know the key:

```python
print(phone_book["Anna"])
```

Python can directly give you:

```text
12345
```

This is similar to opening a telephone book at the letter **A** and looking for **Anna**. But if you only know the phone number:

```text
12345
```

and you want to find the name, it is harder.  You may need to look through the telephone book page by page until you find that number.

A Python dictionary is similar:

> It is designed to find a **value from a key** quickly.

### The main difference

**List**

```text
position → value
```

Example:

```python
cars[2]
```

**Dictionary**

```text
key → value
```

Example:

```python
phone_book["Anna"]
```

For the `mtcars` exercise, the car model can be the key:

```python
cars["Honda Civic"]
```

This means:

> Find `"Honda Civic"` and give me all the information stored for that car.
