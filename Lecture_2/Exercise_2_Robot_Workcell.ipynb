## The robot workcell

The robot starts at **Home** and works with two trays and one assembly fixture.

| Position | Meaning | x-coordinate |
|---|---|---:|
| Home | Robot starting position | `-280` |
| Tray A | Component A | `-180` |
| Tray B | Component B | `-80` |
| Assembly | Assembly fixture | `140` |

The vertical levels are:

| Level | Meaning | y-coordinate |
|---|---|---:|
| Safe level | Horizontal travel | `120` |
| Working level | Pick, place, or fastening action | `-80` |

For the first product, the two fastening points are centered around the assembly position with a spacing of `60`.

The simplified assembly cycle is:

1. transfer Component A from Tray A to the Assembly fixture,
2. transfer Component B from Tray B to the Assembly fixture,
3. fasten the assembly at two positions,
4. return Home.

Moving down at a tray represents **picking** a component.  
Moving down at the assembly fixture represents **placing** or **fastening**.

# Part 1 — Start with a working program

The program below performs one complete assembly cycle.

There are **no user-defined functions yet**.

Copy the program into Spyder, run it, and inspect both the code and the robot path.


````python
import turtle

# Create the drawing window.
screen = turtle.Screen()
screen.setup(width=700, height=450)

# Create the turtle that represents the robot.
robot = turtle.Turtle()
robot.speed(3)



# Workcell positions


# Horizontal positions of the different stations.
home_x = -280
tray_a_x = -180
tray_b_x = -80
assembly_x = 140

# Vertical levels used by the robot.
# The robot travels horizontally at the safe level
# and moves down to the working level to perform an operation.
safe_y = 120
work_y = -80

# Distance between the two fastening points
screw_spacing = 60



# Move the robot to the starting position


# Move to Home without drawing the initial path.
robot.penup()
robot.goto(home_x, safe_y)
robot.pendown()


# Transfer Component A
# --------------------

# Move horizontally to Tray A
robot.goto(tray_a_x, safe_y)

# Move down to pick Component A
robot.goto(tray_a_x, work_y)

# Return to the safe travel level
robot.goto(tray_a_x, safe_y)

# Move horizontally to the assembly fixture.
robot.goto(assembly_x, safe_y)

# Move down to place Component A
robot.goto(assembly_x, work_y)

# Return to the safe travel level.
robot.goto(assembly_x, safe_y)



# Transfer Component B
# --------------------

# Move horizontally to Tray B
robot.goto(tray_b_x, safe_y)

# Move down to pick Component B
robot.goto(tray_b_x, work_y)

# Return to the safe travel level
robot.goto(tray_b_x, safe_y)

# Move horizontally to the assembly fixture
robot.goto(assembly_x, safe_y)

# Move down to place Component B
robot.goto(assembly_x, work_y)

# Return to the safe travel level
robot.goto(assembly_x, safe_y)


# Fasten the assembly at two positions
# --------------------

# Calculate the positions of the two fastening points.
# They are placed equally on each side of the assembly center.
left_x = assembly_x - screw_spacing / 2
right_x = assembly_x + screw_spacing / 2

# Move to the left fastening point.
robot.goto(left_x, safe_y)

# Move down to perform the fastening operation
robot.goto(left_x, work_y)

# Return to the safe travel level
robot.goto(left_x, safe_y)

# Move to the right fastening point
robot.goto(right_x, safe_y)

# Move down to perform the fastening operation
robot.goto(right_x, work_y)

# Return to the safe travel level
robot.goto(right_x, safe_y)


# Return Home
# --------------------


# Return to the robot starting position.
robot.goto(home_x, safe_y)

# Keep the turtle window open.
turtle.done()
````
