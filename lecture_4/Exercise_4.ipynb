## Background: stress, strain, and ultimate tensile strength

When a rod is pulled by a force $F$ along its axis, the **tensile stress** in it is the force divided by the cross-sectional area it acts on:

$$
\Large\boxed{\displaystyle R_m = \frac{F_{\max}}{A_0}}
$$

As you keep increasing the force, the material eventually breaks. The stress at the moment it breaks is called the **ultimate tensile strength (UTS)**, and it is also known as the material's **tensile strength**. For a batch of identical specimens made of the *same* material, this value should come out roughly the same for every specimen, no matter how thick or thin each one is — a thicker specimen just needs a bigger force to reach the same stress. That is exactly the property the two exercises below are
built around.
